```python
import matplotlib.pyplot as plt
import numpy as np

# Example confusion matrix (you can ignore these values if you only want text)
cm = np.array([[50, 10],
               [5,  35]])

# Custom text for each cell (same shape as cm)
labels = np.array([
    ["TP\ntrue positive", "FP\nfalse positive"],
    ["FN\nfalse negative",  "TN\ntrue negative"]
])

fig, ax = plt.subplots()

# Show matrix as image
im = ax.imshow(cm, cmap="Blues")

# Add text annotations
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(j, i, labels[i, j],
                ha="center", va="center",
                color="black", fontsize=12)

# Axis labels
ax.set_xlabel("Predicted label", fontsize=14)
ax.set_ylabel("True label", fontsize=14)

# Tick labels
ax.set_xticks([0, 1])
ax.set_yticks([0, 1])
ax.set_xticklabels(["Negative", "Positive"])
ax.set_yticklabels(["Negative", "Positive"])

# Optional colorbar
plt.colorbar(im)

plt.tight_layout()
plt.show()
```
